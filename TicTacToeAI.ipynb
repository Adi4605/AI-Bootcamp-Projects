```python
import math
```

```python
# define board as a list
board = [' ' for _ in range(9)]
```

```python
def print_board(board):
    print()
    for i in range(3):
        print(" | ".join(board[i*3:(i+1)*3]))
        if i < 2:
            print("-" * 9)
    print()
```

```python
winning_combinations = [
    (0,1,2),(3,4,5),(6,7,8),
    (0,3,6),(1,4,7),(2,5,8),
    (0,4,8),(2,4,6)
]
def check_winner(board):
    for a,b,c in winning_combinations:
        if board[a] == board[b] == board[c] and board[a] != ' ':
            return board[a]
    return None
```

```python
def is_draw(board):
    return ' ' not in board
```

```python
def minimax(board, depth, is_maximizing):
    winner = check_winner(board)

    if winner == 'X':
        return 10 - depth
    if winner == 'O':
        return depth - 10
    if is_draw(board):
        return 0

    # Player's turn (maximize)
    if is_maximizing:
        best_score = -math.inf
        for i in range(9):
            if board[i] == ' ':
                board[i] = 'X'
                score = minimax(board, depth + 1, False)
                board[i] = ' '
                best_score = max(best_score, score)
        return best_score

    # AI's turn (minimize)
    else:
        best_score = math.inf
        for i in range(9):
            if board[i] == ' ':
                board[i] = 'O'
                score = minimax(board, depth + 1, True)
                board[i] = ' '
                best_score = min(best_score, score)
        return best_score
```

```python
def find_best_move(board):
    best_score = math.inf
    best_move = -1

    for i in range(9):
        if board[i] == ' ':
            board[i] = 'O'
            score = minimax(board, 0, True)
            board[i] = ' '
            if score < best_score:
                best_score = score
                best_move = i
    return best_move
```

```python
print("Board positions are:")
print("1 | 2 | 3")
print("---------")
print("4 | 5 | 6")
print("---------")
print("7 | 8 | 9")

print_board(board)

def reset_board():
    return [' '] * 9
board = reset_board()

while True:
#Player
    while True:
      try:
          player_move = int(input("Enter your move (1-9): "))
          if player_move < 1 or player_move > 9:
              print("Please enter a number between 1 and 9.")
              continue
          player_move -= 1
          if board[player_move] != ' ':
              print("That position is already occupied.")
              continue
          break
      except ValueError:
          print("Please enter a valid number.")

    board[player_move] = 'X'
    print_board(board)
    winner = check_winner(board)

    if winner == 'X':
        print("Congratulations! You Win!")
        break
    if is_draw(board):
        print("It's a Draw!")
        break
#AI
    print("AI is thinking...\n")
    ai_move = find_best_move(board)
    board[ai_move] = 'O'
    print_board(board)
    winner = check_winner(board)

    if winner == 'O':
        print("AI Wins!")
        break
    if is_draw(board):
        print("It's a Draw!")
        break
```
